```python
import pandas as pd

# Step 1: Load both CSV files
df1 = pd.read_csv('zoom_live_2025-10-27-Part1.csv')
df2 = pd.read_csv('zoom_live_2025-10-27-Part2.csv')

# Step 2: Normalize column names
df1.columns = df1.columns.str.strip()
df2.columns = df2.columns.str.strip()

print("df1 columns:", df1.columns.tolist())
print("df2 columns:", df2.columns.tolist())

# Step 3: Combine both DataFrames
combined = pd.concat([df1, df2], ignore_index=True)

# Step 4: Group by identifying columns (Name + Email) and sum durations
aggregated = (
    combined
    .groupby(['Name (original name)', 'Email', 'Guest'], as_index=False)
    .agg({'Total duration (minutes)': 'sum'})
)

# Step 5: Save full result
aggregated.to_csv('combined_duration.csv', index=False)
print("✅ Saved to combined_duration.csv")

aggregated
```
